```python
import tkinter as tk
from datetime import datetime
import random

root = tk.Tk()
root.title("FocusClock")
root.geometry("650x560")
root.configure(bg="black")

running=False
mode="FOCUS"
time_left=0
theme_index=0

canvas=tk.Canvas(root,width=610,height=380,bg="black",highlightthickness=0)
canvas.pack(pady=10)

clock_text=canvas.create_text(305,60,text="",fill="white",font=("Helvetica",36,"bold"))
mode_text=canvas.create_text(305,110,text="FOCUS",fill="white",font=("Helvetica",18))
timer_text=canvas.create_text(305,260,text="25:00",fill="white",font=("Helvetica",64,"bold"))

bg_items=[]

themes=[
("#38bdf8","OCEAN"),
("#22c55e","FOREST"),
("#facc15","SUN"),
("#a78bfa","GALAXY"),
("#fb7185","DREAM"),
("#10b981","ZEN"),
("#f97316","FIRE"),
("#67e8f9","ICE"),
("#4ade80","NATURE"),
("#e879f9","MAGIC"),
("#f43f5e","LOVE"),
("#fbbf24","ENERGY")
]

def clear_bg():
    for i in bg_items:
        canvas.delete(i)
    bg_items.clear()

def draw_theme():
    clear_bg()
    color,word=themes[theme_index]

    canvas.configure(bg="black")

    bg_items.append(
        canvas.create_text(
            305,190,
            text=word,
            fill=color,
            font=("Helvetica",96,"bold")
        )
    )

    if word=="OCEAN":
        for i in range(6):
            bg_items.append(canvas.create_arc(50+i*90,300,150+i*90,360,start=0,extent=180,style="arc",outline=color,width=3))

    elif word=="FOREST":
        for i in range(5):
            bg_items.append(canvas.create_polygon(80+i*100,320,110+i*100,240,140+i*100,320,fill=color))

    elif word=="SUN":
        bg_items.append(canvas.create_oval(270,150,340,220,fill=color,outline=""))

    elif word=="GALAXY":
        for i in range(40):
            x=random.randint(0,600)
            y=random.randint(0,380)
            bg_items.append(canvas.create_oval(x,y,x+3,y+3,fill=color,outline=""))

    elif word=="DREAM":
        for i in range(7):
            bg_items.append(canvas.create_oval(70+i*80,290,130+i*80,340,fill=color,outline=""))

    elif word=="ZEN":
        bg_items.append(canvas.create_oval(240,210,370,340,outline=color,width=5))

    elif word=="FIRE":
        for i in range(6):
            bg_items.append(canvas.create_polygon(80+i*80,340,100+i*80,260,120+i*80,340,fill=color))

    elif word=="ICE":
        for i in range(6):
            bg_items.append(canvas.create_rectangle(70+i*90,290,110+i*90,340,fill=color))

    elif word=="NATURE":
        for i in range(6):
            bg_items.append(canvas.create_oval(70+i*90,310,120+i*90,350,fill=color))

    elif word=="MAGIC":
        for i in range(20):
            x=random.randint(0,600)
            y=random.randint(200,360)
            bg_items.append(canvas.create_text(x,y,text="✦",fill=color,font=("Helvetica",16)))

    elif word=="LOVE":
        for i in range(8):
            bg_items.append(canvas.create_text(80+i*70,320,text="❤",fill=color,font=("Helvetica",24)))

    elif word=="ENERGY":
        for i in range(6):
            bg_items.append(canvas.create_line(80+i*90,260,100+i*90,340,fill=color,width=4))

def update_clock():
    canvas.itemconfig(clock_text,text=datetime.now().strftime("%H:%M:%S"))
    root.after(1000,update_clock)

def tick():
    global running,time_left,mode

    if running and time_left>0:
        time_left-=1

    if running and time_left==0:
        if mode=="FOCUS":
            mode="RELAX"
            time_left=int(relax_entry.get())*60
        else:
            mode="FOCUS"
            time_left=int(focus_entry.get())*60

    m,s=divmod(time_left,60)

    canvas.itemconfig(timer_text,text=f"{m:02d}:{s:02d}")
    canvas.itemconfig(mode_text,text=mode)

    root.after(1000,tick)

def start():
    global running,time_left
    if time_left==0:
        time_left=int(focus_entry.get())*60
    running=True

def pause():
    global running
    running=False

def reset():
    global running,time_left,mode
    running=False
    mode="FOCUS"
    time_left=int(focus_entry.get())*60

def change_theme():
    global theme_index
    theme_index=(theme_index+1)%len(themes)
    draw_theme()

control=tk.Frame(root,bg="black")
control.pack()

tk.Label(control,text="Focus",fg="white",bg="black",font=("Helvetica",12)).grid(row=0,column=0)
focus_entry=tk.Entry(control,width=6,justify="center",font=("Helvetica",12))
focus_entry.insert(0,"25")
focus_entry.grid(row=1,column=0,padx=10)

tk.Label(control,text="Relax",fg="white",bg="black",font=("Helvetica",12)).grid(row=0,column=1)
relax_entry=tk.Entry(control,width=6,justify="center",font=("Helvetica",12))
relax_entry.insert(0,"5")
relax_entry.grid(row=1,column=1,padx=10)

buttons=tk.Frame(root,bg="black")
buttons.pack(pady=20)

def btn(name,cmd,color):
    return tk.Button(
        buttons,
        text=name,
        command=cmd,
        width=12,
        height=2,
        bg=color,
        fg="black",
        font=("Helvetica",14,"bold"),
        relief="flat",
        activebackground="white"
    )

btn("START",start,"#4ade80").grid(row=0,column=0,padx=10)
btn("PAUSE",pause,"#facc15").grid(row=0,column=1,padx=10)
btn("RESET",reset,"#f87171").grid(row=0,column=2,padx=10)
btn("THEME",change_theme,"#38bdf8").grid(row=0,column=3,padx=10)

draw_theme()
update_clock()
tick()

root.mainloop()
```
